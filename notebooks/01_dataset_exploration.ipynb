```python
from datasets import load_dataset

dataset = load_dataset("squad")
print(dataset)
```

```python
sample = dataset["train"][0]
print("Question:", sample["question"])
print("\nContext preview:", sample["context"][:700])
print("\nAnswers:", sample["answers"])
```

## Key fields
- `context`: paragraph containing the answer.
- `question`: question about the paragraph.
- `answers.text`: reference answer text.
- `answers.answer_start`: character offset where the answer begins.
